```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/Sample - Superstore.csv',encoding='latin1')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
total_sales = df['Sales'].sum()
print("Total Sales:",total_sales)
```

```python
total_profit = df['Profit'].sum()
print("Total Profit:",total_profit)
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
sales_region = df.groupby('Region')['Sales'].sum()

print(sales_region)
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

print(category_sales)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Month'] = df['Order Date'].dt.month

monthly_sales = df.groupby('Month')['Sales'].sum()

print(monthly_sales)
```

```python
top_customers = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)

print(top_customers)
```

```python
top_products.plot(kind='bar',figsize=(12,8))

plt.title('Top 10 products by sales')
plt.xlabel('Product Name')
plt.ylabel('Sales')

plt.xticks(rotation=75)

plt.show()
```

```python
sales_region.plot(kind = 'bar',figsize=(8,5))

plt.title('Sales by region')
plt.xlabel('Region')
plt.ylabel('Sales')

plt.show()
```

```python
category_sales.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7))

plt.title('Category-wise Sales')

plt.ylabel('')

plt.show()
```

```python
monthly_sales.plot(kind='line',marker='o',figsize=(10,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.grid(True)

plt.show()
```

```python
top_customers.plot(kind='bar',figsize=(12,8))

plt.title('Top Customers by Sales')
plt.xlabel('Customer Name')
plt.ylabel('Sales')

plt.xticks(rotation=75)

plt.show()
```
